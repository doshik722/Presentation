# запускаем проект локально или в Kaggle
переносим **весь проект**, включая `src`, `configs` и `data`. В Kaggle подключаем его как Dataset и указываем папку с `pyproject.toml`. Держим алгоритмы и метрики в Python-модулях. По умолчанию проверяем только план: веса не скачиваются, генерация не запускается.

Кодовые ячейки удалены. Ниже сохранены заметки к будущей реализации; запуск пока недоступен.


Не перезаписываем уже существующую рабочую копию. Для новой версии Dataset выбираем новое имя `WORKING_PROJECT`, чтобы не смешать версии кода. Сохраняем результаты из `/kaggle/working` через output версии ноутбука.

Для первой установки меняем `INSTALL_DEPENDENCIES` на `True`. Перед реальной генерацией выбираем `EXTRAS = 'models'`; для 4-bit — `'models,quantization'`. Проверяем совместимость окружения отдельно и перезапускаем kernel, если этого требует установка. Для загрузки весов и пакетов нам нужен доступ к интернету либо заранее подготовленные локальные файлы.
